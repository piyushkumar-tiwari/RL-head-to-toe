## Matrix Multiplication without numpy

#### 1. Simple Matrix Multplication with using predefined matrices.

```python
a = [[0,1,2],[5,6,7],[77,88,99],[6.5,4.5,8.5]]
b= [[-10,7], [6.7,8], [7,9]]

i = len(a)
j= len(a[0])
x = len (b)
y = len(b[0])

if j!=x:
    print("ERROR!")
else:
    z = [[0,0],[0,0],[0,0],[0,0]]
    for i in range(i):
        for j in range(y):
            for k in range(x):
                z[i][j] += a[i][k] * b[k][j]
    print('Matrix multiplcation gives us:', z)
```

## Matrix Multiplication with Numpy

```python
import numpy as np

a = input("Enter the dimensions of matrix1:")
a = a.replace(',',' ')
a= a.replace('x', ' ')
a= a.replace('X', ' ')
a= a.replace('*', ' ')
a = a.split()
i = int(a[0])
j = int(a[1])

matrix1 = np.empty((i,j))
for rows in range(i):
    for cols in range(j):
        matrix1[rows,cols] = float(input(f'Please enter the element [{rows},{cols}] in matrix1:'))
print('The dimensions of matrix 1 are:', i, 'x', j)
print('The input matrix 1 by the user is:',matrix1)

b= input('Please enter the dimensions of matrix2:')
b = b.replace(',', ' ')
b = b.replace('x', ' ')
b= b.replace ('X', ' ')
b= b.replace('*', ' ')
b= b.split()
x= int(b[0])
y= int(b[1])
matrix2 = np.empty((x,y))
for rows in range(x):
    for cols in range(y):
        matrix2[rows,cols] = float(input(f'Please enter the element [{rows},{cols}] in matrix2:'))
print('The dimensions of matrix 2 are:', x, 'x', y)
print('The input matrix 2 by the user is:',matrix2)

if j!=x:
    print("Error. No matrix multiplication is possible in this case. Ensure columns in matrix 1 = Rows in matrix 2")
else:
    z = matrix1 @ matrix2
    print('The output matrix is:',z)
```

```python
#Writing the code calculating the formula for probability distribution across various actions using LSE

import numpy as np

class RL_Softmax:
    def __init__(self, input_array):
        self.data = np.array(input_array)
        sum = 0

    def LSE (self):
        x = np.max(self.data)
        e_i = self.data - x
        e_i = np.exp(e_i)
        e_j = np.sum(e_i)
        return x+ (np.log(e_j))
    def get_probability(self):
        LSE_scalar = self.LSE()
        theta = self.data-LSE_scalar
        return np.exp(theta)
    def sum_of_probabilities(self):
        sum = self.get_probability()
        return np.sum(sum)
    
q_values = np.array([1000, 1001, 1002])

sf= RL_Softmax(q_values)
result = sf.LSE()
print("LSE Scalar is:",result)
probability = sf.get_probability()
print("The Valid probability distribution is:", probability)
sum_of_probability = sf.sum_of_probabilities()
print('Sum of probabilities is:', sum_of_probability)
```

```python
#Combining the concepts of max q value extraction and calculating valid probability distribution from scratch!

import numpy as np

class stable_softmax:
    def __init__(self,input_array):
        self.data = np.array(input_array)
    def max_extractor(self):
        y = np.max(self.data, axis = 2, keepdims=True)
        return y
    def exponential(self):
        y = self.max_extractor()
        exp = self.data - y
        exp = np.exp(exp)
        exp = np.sum(exp, axis = 2, keepdims=True)
        return np.log(exp) + (y)
    def get_probability(self):
        k = self.data
        j = self.exponential()
        a = k-j
        return np.exp(a)
    
data = np.random.randint(700, 950, (5, 85, 4))
k = stable_softmax(data)
print(k.get_probability())
```
