## Matrix Multiplication without numpy

#### 1. Simple Matrix Multplication with using predefined matrices.

```python
a = [[0,1,2],[5,6,7],[77,88,99],[6.5,4.5,8.5]]
b= [[-10,7], [6.7,8], [7,9]]

i = len(a)
j= len(a[0])
x = len (b)
y = len(b[0])

if j!=x:
    print("ERROR!")
else:
    z = [[0,0],[0,0],[0,0],[0,0]]
    for i in range(i):
        for j in range(y):
            for k in range(x):
                z[i][j] += a[i][k] * b[k][j]
    print('Matrix multiplcation gives us:', z)
```

## Matrix Multiplication with Numpy

```python
import numpy as np

a = input("Enter the dimensions of matrix1:")
a = a.replace(',',' ')
a= a.replace('x', ' ')
a= a.replace('X', ' ')
a= a.replace('*', ' ')
a = a.split()
i = int(a[0])
j = int(a[1])

matrix1 = np.empty((i,j))
for rows in range(i):
    for cols in range(j):
        matrix1[rows,cols] = float(input(f'Please enter the element [{rows},{cols}] in matrix1:'))
print('The dimensions of matrix 1 are:', i, 'x', j)
print('The input matrix 1 by the user is:',matrix1)

b= input('Please enter the dimensions of matrix2:')
b = b.replace(',', ' ')
b = b.replace('x', ' ')
b= b.replace ('X', ' ')
b= b.replace('*', ' ')
b= b.split()
x= int(b[0])
y= int(b[1])
matrix2 = np.empty((x,y))
for rows in range(x):
    for cols in range(y):
        matrix2[rows,cols] = float(input(f'Please enter the element [{rows},{cols}] in matrix2:'))
print('The dimensions of matrix 2 are:', x, 'x', y)
print('The input matrix 2 by the user is:',matrix2)

if j!=x:
    print("Error. No matrix multiplication is possible in this case. Ensure columns in matrix 1 = Rows in matrix 2")
else:
    z = matrix1 @ matrix2
    print('The output matrix is:',z)
```

Without numpy:

```python
import math

class softmax:
    def __init__(self, input_array):
        self.data = input_array
        sum = 0
        for i in self.data:
            e_i = math.exp(i)
            sum += e_i
        
        if sum == 0:
            ("Error. NaN")
        else:
            for j in self.data:
                e_j = math.exp(j)
                e_jf = e_j/sum 
                print(e_jf)
data1 = [1, 10, 50]
data2 = [1000, 1002, 1010]
sm1 = softmax(data1)
print(sm1)
sm2 = softmax(data2)
print(sm2)
```

```python
import math

class custom_softmax:
    def __init__(self, input_array):
        self.data = input_array
        x_max = max(self.data)
        sum = 0
        for i in self.data:
            e_i = math.exp(i-x_max)
            sum += e_i
        for j in self.data:
            e_j = math.exp(j-x_max)
            e_jf = e_j/sum
            print(e_jf)

data1 = [1, 10, 50]
data2 = [1000, 1002, 1010]
sm1 = custom_softmax(data1)
print(sm1)
sm2 = custom_softmax(data2)
print(sm2)
```
