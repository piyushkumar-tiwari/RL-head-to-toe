## ***Environment Definition*** ##

```python
import numpy as np

class Airplane:
    def __init__(self, K = 25 ,T = 10, prices = [50, 100, 150, 200, 250]):
        self.time = T
        self.prices = prices
        self.seats_remaining = K
        self.reset()

    def reset(self):
        self.time_left = self.time
        self.seats_remain = self.seats_remaining
        return self._get_state_index()

    def _get_state_index(self):
        return self.seats_remain * (self.T + 1) + self.time_left
    
    def step (self, action_index):
        price = self.prices[action_index]
        avg_demand = 250/price
        demand = np.random.poisson(avg_demand)
        seats_sold = min(self.seats_remaining, demand)
        reward = seats_sold * price
        self.current_seats -= seats_sold
        self.days_left -= 1

        done = (self.days_left == 0) or (self.current_seats == 0)
        return self._get_state_index(), reward, done
```

### ***SARSA*** ###

```python
import numpy as np

class SARSA:
    def __init__(self,  num_episodes, states, actions, alpha = 0.9, gamma = 0.5, epsilon = 0.1):
        self.states = np.arange(0, states)
        self.actions = np.arange(0,actions)
        self.alpha = alpha
        self.gamma = gamma
        self.epsilon = epsilon
        self.num_episodes = num_episodes
        self.q_table = np.zeros((len(self.states), len(self.actions)))

    def choose_action(self, state_index):
        if np.random.uniform(0,1) < self.epsilon:
            return np.random.choice(self.actions)
        else:
            return np.argmax(self.q_table[state_index])
    
    def doSARSA(self, state, action, next_state, next_action, reward):
        new_q = self.q_table[state, action] + self.alpha * (reward + self.gamma * self.q_table[next_state, next_action] - self.q_table[state, action])
        self.q_table[state, action] = new_q
        return self.q_table
    
    def train(self, env, num_episodes):
        for e in range(self.num_episodes):
            state = env.reset()
            action = self.choose_action(state)
            done = False
            while not done:
                next_state, reward, done = env.step(action)
                next_action = self.choose_action(next_state)

                self.doSARSA(state, action, next_state, next_action, reward)
                state = next_state
                action = next_action
            self.epsilon *= 0.99
```

### ***Q-Learning*** ###

```python
import numpy as np

class QLearning:
    def __init__(self, states, actions, num_episodes, epsilon = 0.1, gamma = 0.9, beta = 0.5):
        self.states = np.arange(0, states + 1)
        self.actions = np.arange(0, actions+1)   
        self.num_episodes = num_episodes
        self.epsilon = epsilon
        self.beta = beta
        self.gamma = gamma

        self.q_table = np.zeros(shape=(self.states.size,self.actions.size))

    def choose_action(self, state_index):
        if np.random.uniform(0,1) < self.epsilon:
            return np.random.choice(self.actions)
        else:
            return np.argmax(self.q_table[state_index])
        
    def step (self, action, state, reward, next_state):
        new_q = self.q_table[state,action] + self.beta * (reward + (self.gamma * np.max(self.q_table[next_state])) - self.q_table[state, action])
        self.q_table[state,action] = new_q
```
